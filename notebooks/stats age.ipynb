```python
# Age Statistical Analysis

## Purpose

This notebook performs statistical analysis of resident age in the StanzaOps hostel dataset.

## Main Tasks

- Calculate age statistics
- Analyse age distribution
- Identify minimum and maximum age
- Check for age outliers
- Generate age-related insights

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\projects\StanzaOps-Hostel-Operations-Revenue-Intelligence-Engine\data\processed\Hostel_Management_Dataset_Cleaned.csv")
```

```python
df[["Age", "Deposit_Amount", "Amount_Paid"]].describe()
```

```python
df[["Age", "Deposit_Amount", "Amount_Paid"]].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Age Skewness:", df["Age"].skew())
print("Age Kurtosis:", df["Age"].kurt())
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Deposit_Amount"], bins=10)
plt.title("Deposit Amount Distribution")
plt.xlabel("Deposit Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Deposit Skewness:", df["Deposit_Amount"].skew())
print("Deposit Kurtosis:", df["Deposit_Amount"].kurt())
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Amount_Paid"], bins=10)
plt.title("Amount Paid Distribution")
plt.xlabel("Amount Paid")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Amount Paid Skewness:", df["Amount_Paid"].skew())
print("Amount Paid Kurtosis:", df["Amount_Paid"].kurt())
```

```python
1. This dataset is a synthetic hostel management dataset created for learning and portfolio purposes.
2. Deposit_Amount is fixed at ₹10,000 for all residents, so there is no variation in this feature.
3. Amount_Paid contains limited values based on payment status (Paid and Partial), which may not fully represent real-world payment patterns.
4. The dataset contains balanced records across genders and cities, which may differ from actual hostel populations.
5. The findings from this analysis are based on this sample dataset and should not be generalized to all hostel management systems.
```
