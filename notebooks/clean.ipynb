```python
# Data Cleaning

## Purpose

This notebook performs cleaning and preprocessing of the StanzaOps hostel dataset.

## Main Tasks

- Identify data quality issues
- Handle missing values
- Check duplicate records
- Standardize categorical and date fields
- Prepare data for analysis 
```

```python
import pandas as pd

df = pd.read_csv(
    r"D:\projects\StanzaOps-Hostel-Operations-Revenue-Intelligence-Engine\Copy of HostelManagementDatasetCleaned.csv.csv"
)
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["Remarks"] = df["Remarks"].fillna("No Remarks")
```

```python
print("Duplicate Resident_ID:", df["Resident_ID"].duplicated().sum())
print("Duplicate Mobile_Number:", df["Mobile_Number"].duplicated().sum())
```

```python
assert df.shape[0] == 10000, "Row count does not match!"
print("Row count is correct.")
```

```python
assert df["Resident_ID"].duplicated().sum() == 0, "Duplicate Resident_ID found!"
assert df["Mobile_Number"].duplicated().sum() == 0, "Duplicate Mobile_Number found!"

print("Duplicate validation passed.")
```

```python
df.isnull().sum()
```

```python
assert df["Remarks"].isnull().sum() == 0, "Remarks still contains null values!"

print("Null validation passed.")
```

```python
df.to_csv(
    r"D:\projects\StanzaOps-Hostel-Operations-Revenue-Intelligence-Engine\data\processed\Hostel_Management_Dataset_Cleaned.csv",
    index=False
)
```
