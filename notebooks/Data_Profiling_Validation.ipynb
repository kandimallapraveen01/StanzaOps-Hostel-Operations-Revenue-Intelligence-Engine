```python
# Data Profiling and Validation

## Purpose

This notebook profiles and validates the StanzaOps hostel dataset to assess data quality and reliability.

## Main Tasks

- Check dataset dimensions
- Analyse data types
- Check missing values
- Check unique values
- Validate duplicate records
- Validate important business rules
import pandas as pd
```

```python
df = pd.read_csv("Hostel_Management_Dataset_Filled.csv")
print(df.shape)
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
duplicate_resident = df.duplicated(subset=['Resident_ID']).sum()
print("Duplicate Resident_ID:", duplicate_resident)
```

```python
duplicate_mobile = df.duplicated(subset=['Mobile_Number']).sum()
print("Duplicate Mobile_Number:", duplicate_mobile)
```
